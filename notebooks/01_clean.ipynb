```python
def job_clean(x):
    if pd.isnull(x):
        return 'Not Specified'
    x=x.lower()
    if "intern" in x:
        return "Internship"
    elif "contract" in x or "temp" in x:
        return "Contract/Temp"
    elif "part-time" in x:
        return "Part-time"
    elif "full-time" in x or "permanent" in x:
        return "Full-time"
    else:
        return "Other"
df_clean['job_type_clean']=df_clean['job_type'].apply(job_clean)
df_clean['job_type_clean'].value_counts()

```

```python
df_clean['job_title'].isnull().sum()
```

```python
df_clean['job_title'].value_counts()
```

```python
def job_title_clean(x):
    if pd.isnull(x):
        return 'Not Specified'
    x=x.lower()
    if ('sr.' in x or 'sr ' in x or 'senior' in x) and 'data' in x:
        return 'Senior Data Analyst'
    elif 'business' in x and 'intelligence' in x:
        return 'BI Analyst'
    elif 'data' in x:
        return 'Data Analyst'
    elif 'business' in x:
        return 'Business Analyst'
    else:
        return 'Others'

df_clean['job_title_cleaned']=df_clean['job_title'].apply(job_title_clean)
df_clean['job_title_cleaned'].value_counts()
       
    
```
